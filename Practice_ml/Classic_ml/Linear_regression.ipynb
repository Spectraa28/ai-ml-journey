# LINEAR REGRESSION FROM SCCRATCH WITH GRADIENT DESCENT 

```python
class LinearRegression:
    def __init__(self,learning_rate=0.01):
        self.w = 0.0
        self.b = 0.0
        self.lr = learning_rate
        self.cost_history = []
        
    def predict(self,X):
        return [self.w  * x + self.b for x in X]
    
    def compute_cost(self,X,y):
        prediction = self.predict(X)
        n = len(y)
        cost = sum((pred - actual)**2 for pred,actual in zip(prediction,y)) / n
        return cost
    
    def compute_gradient(self,X,y):
        prediction = self.predict(X)
        n=len(y)
        dw = (2/n) * sum((pred-actual) * x for pred , actual ,x in zip(prediction,y,X))
        db =  (2/n) * sum((pred,actual) for pred,actual in zip(prediction,y))
        
    def fit(self,X,y,epochs=1000):
        for epoch in range(epochs):
            dw,db = self.compute_gradient(X,y)
            self.w -= self.lr * dw
            self.b -= self.lr *db
            cost = self.compute_cost(X,y)
            self.cost_history.append(cost)
            return self
        
    def r_squareed(self,X,y):
        prediction = self.predict(X)
        y_mean = sum(y) / len(y)
        ss_res = sum((actual , pred) ** 2 for actual , pred in zip(y,prediction))
        ss_tot = sum((actual - y_mean)**2 for actual , y_mean in zip(y,y_mean))
        return 1 - (ss_res/ss_tot)
        
```
